# Model Training

## Objective


This notebook trains multiple regression models.

The project predicts:

Next Cycle Length
Next Period Length

# Import Libraries

```python
import warnings
warnings.filterwarnings("ignore")

import joblib

from sklearn.linear_model import LinearRegression

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from sklearn.neighbors import KNeighborsRegressor

from sklearn.svm import SVR

from xgboost import XGBRegressor
```

# Load Processed Data

# cycle and period model

```python
X_train = joblib.load("../processed_data/Xc_train.pkl")
X_valid = joblib.load("../processed_data/Xc_valid.pkl")
X_test = joblib.load("../processed_data/Xc_test.pkl")

y_train = joblib.load("../processed_data/yc_train.pkl")
y_valid = joblib.load("../processed_data/yc_valid.pkl")
y_test = joblib.load("../processed_data/yc_test.pkl")
```

```python
Xp_train = joblib.load("../processed_data/Xp_train.pkl")
Xp_valid = joblib.load("../processed_data/Xp_valid.pkl")
Xp_test = joblib.load("../processed_data/Xp_test.pkl")

yp_train = joblib.load("../processed_data/yp_train.pkl")
yp_valid = joblib.load("../processed_data/yp_valid.pkl")
yp_test = joblib.load("../processed_data/yp_test.pkl")
```

# Create Models

```python
models = {

    "Linear Regression":
        LinearRegression(),

    "Decision Tree":
        DecisionTreeRegressor(
            random_state=42
        ),

    "Random Forest":
        RandomForestRegressor(
            random_state=42
        ),

    "KNN":
        KNeighborsRegressor(),

    "SVR":
        SVR(),

    "Gradient Boosting":
        GradientBoostingRegressor(
            random_state=42
        ),

    "XGBoost":
        XGBRegressor(
            random_state=42,
            objective="reg:squarederror"
        )
}
```

# Train Cycle Models

```python
trained_cycle_models = {}

for name, model in models.items():

    print(f"Training {name}")

    model.fit(
        X_train,
        y_train
    )

    trained_cycle_models[name] = model
```

# Save Cycle Models

```python
from pathlib import Path

output_dir = Path("../models/trained_models")
output_dir.mkdir(parents=True, exist_ok=True)

for name, model in trained_cycle_models.items():

    filename = output_dir / f"{name.replace(' ', '_')}_cycle.pkl"

    joblib.dump(
        model,
        filename
    )
```

# Train Period Models

```python
trained_period_models = {}

for name, model in models.items():

    print(f"Training {name}")

    model.fit(
        Xp_train,
        yp_train
    )

    trained_period_models[name] = model
```

# Save Period Models

```python
for name, model in trained_period_models.items():

    filename = (
        "../models/trained_models/"
        + name.replace(" ", "_")
        + "_period.pkl"
    )

    joblib.dump(
        model,
        filename
    )
```
